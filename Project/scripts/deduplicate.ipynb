```python
import pandas as pd
import numpy as np

# 1. Load CSV dataset
CSV_DATA_PATH = r"C:\Project\data\raw\flower_prices_with_festivals_2010_2016.csv"
OUTPUT_CSV_PATH = r"C:\Project\data\processed\clean_prices_festivals_2010_2016.csv"

df = pd.read_csv(CSV_DATA_PATH)
df['DATE'] = pd.to_datetime(df['DATE'])

# 2. Group by DATE to merge co-occurring festival names and preserve single rows
df_dedup = df.groupby('DATE').agg({
    'festival_name': lambda x: ' | '.join(sorted(x.dropna().astype(str).unique())) if len(x.dropna()) > 0 else np.nan,
    'tithi': lambda x: ' | '.join(sorted(x.dropna().astype(str).unique())) if len(x.dropna()) > 0 else np.nan,
    'hindu_month': lambda x: ' | '.join(sorted(x.dropna().astype(str).unique())) if len(x.dropna()) > 0 else np.nan,
    'price': 'first'  # Price is identical across co-occurring festival entries
}).reset_index()

# 3. Add festival binary flag
df_dedup['is_festival'] = df_dedup['festival_name'].notna().astype(int)

# 4. Save the fully deduplicated dataset
df_dedup.to_csv(OUTPUT_CSV_PATH, index=False)
print(f"[SUCCESS]: Saved cleaned dataset to {OUTPUT_CSV_PATH} ({len(df_dedup)} rows)\n")


# Create display dataframe with requested column names
df_display = df_dedup.rename(columns={
    'DATE': 'Date',
    'festival_name': 'Co-occurring Festivals',
    'price': 'Price (₹)'
})

# Format Date to YYYY-MM-DD
df_display['Date'] = df_display['Date'].dt.strftime('%Y-%m-%d')

# Filter for rows where multiple (2) festivals co-exist
df_co_occurring = df_display[df_display['Co-occurring Festivals'].str.contains('\|', na=False)]

print("--- CO-OCCURRING FESTIVALS DEDUPLICATED TABLE ---")
print(df_co_occurring[['Date', 'Co-occurring Festivals', 'Price (₹)']].to_string(index=False))
```
